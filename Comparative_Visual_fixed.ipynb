```python

import pandas as pd
import plotly.graph_objects as go
from IPython.display import display

# ------------------------
# DATA
# ------------------------
time_series_data = [
    {'quarter': '2011-Q1', 'gdp': 2102863, 'gva': 1969132},
    {'quarter': '2011-Q4', 'gdp': 2365461, 'gva': 2150712},
    {'quarter': '2012-Q3', 'gdp': 2344767, 'gva': 2177528},
    {'quarter': '2013-Q2', 'gdp': 2357079, 'gva': 2193897},
    {'quarter': '2014-Q1', 'gdp': 2535751, 'gva': 2377154},
    {'quarter': '2014-Q4', 'gdp': 2783733, 'gva': 2498612},
    {'quarter': '2015-Q3', 'gdp': 2836387, 'gva': 2637004},
    {'quarter': '2016-Q2', 'gdp': 3035756, 'gva': 2791994},
    {'quarter': '2017-Q1', 'gdp': 3146336, 'gva': 2938107},
    {'quarter': '2017-Q4', 'gdp': 3516058, 'gva': 3116598},
    {'quarter': '2018-Q3', 'gdp': 3489761, 'gva': 3193097},
    {'quarter': '2019-Q2', 'gdp': 3549945, 'gva': 3259776},
    {'quarter': '2020-Q1', 'gdp': 2732306, 'gva': 2607752},
    {'quarter': '2020-Q4', 'gdp': 3952568, 'gva': 3592345},
    {'quarter': '2021-Q3', 'gdp': 3870108, 'gva': 3569131},
    {'quarter': '2022-Q2', 'gdp': 3891251, 'gva': 3582738},
    {'quarter': '2023-Q1', 'gdp': 4170114, 'gva': 3892452},
    {'quarter': '2023-Q4', 'gdp': 4782032, 'gva': 4285957},
    {'quarter': '2024-Q3', 'gdp': 4716992, 'gva': 4312585},
]

df_time = pd.DataFrame(time_series_data)

model_performance_data = [
    {'name': 'Linear Regression', 'mape': 3.15},
    {'name': 'Polynomial (d=2)', 'mape': 2.54},
    {'name': 'Polynomial (d=3)', 'mape': 2.41},
    {'name': 'ARIMA', 'mape': 2.06},
    {'name': 'SARIMA', 'mape': 1.23},
]
df_model = pd.DataFrame(model_performance_data)

forecast_data = [
    {'quarter': '2023-Q4', 'actual': 4782032, 'predicted': 4730000},
    {'quarter': '2024-Q1', 'actual': 4441986, 'predicted': 4390000},
    {'quarter': '2024-Q2', 'actual': 4492230, 'predicted': 4450000},
    {'quarter': '2024-Q3', 'actual': 4716992, 'predicted': 4680000},
    {'quarter': '2024-Q4', 'predicted': 4852147},
    {'quarter': '2025-Q1', 'predicted': 4523916},
    {'quarter': '2025-Q2', 'predicted': 4589244},
    {'quarter': '2025-Q3', 'predicted': 4792381},
]
df_forecast = pd.DataFrame(forecast_data)

seasonal_data = [
    {'name': 'Q1', 'avgGDP': 3390130},
    {'name': 'Q2', 'avgGDP': 3550520},
    {'name': 'Q3', 'avgGDP': 3650785},
    {'name': 'Q4', 'avgGDP': 3942647},
]
df_seasonal = pd.DataFrame(seasonal_data)

transferability_data = [
    {'model': 'Linear Regression', 'gdp': 3.15, 'gva': 3.04},
    {'model': 'Polynomial (d=2)', 'gdp': 2.54, 'gva': 2.48},
    {'model': 'ARIMA', 'gdp': 2.06, 'gva': 1.98},
    {'model': 'SARIMA', 'gdp': 1.23, 'gva': 1.19},
]
df_transfer = pd.DataFrame(transferability_data)

# ------------------------
# FIGURE 1: GDP & GVA TREND
# ------------------------
fig1 = go.Figure()
fig1.add_trace(go.Scatter(
    x=df_time['quarter'], y=df_time['gdp'],
    mode='lines+markers', name='GDP'
))
fig1.add_trace(go.Scatter(
    x=df_time['quarter'], y=df_time['gva'],
    mode='lines+markers', name='GVA'
))
fig1.update_layout(
    title="Figure 1: GDP and GVA Time Series (2011–2024)",
    xaxis_title="Quarter",
    yaxis_title="Value (INR)",
    template="plotly_white"
)
fig1.show()

# ------------------------
# FIGURE 2: MODEL PERFORMANCE
# ------------------------
fig2 = go.Figure()
fig2.add_trace(go.Bar(
    x=df_model['name'], y=df_model['mape'], name='MAPE'
))
fig2.update_layout(
    title="Figure 2: Model Performance Comparison (MAPE)",
    xaxis_title="Model",
    yaxis_title="MAPE (%)",
    template="plotly_white"
)
fig2.show()

# ------------------------
# FIGURE 3: FORECAST RESULTS
# ------------------------
fig3 = go.Figure()
fig3.add_trace(go.Scatter(
    x=df_forecast['quarter'], y=df_forecast['actual'],
    mode='lines+markers', name='Actual GDP'
))
fig3.add_trace(go.Scatter(
    x=df_forecast['quarter'], y=df_forecast['predicted'],
    mode='lines+markers', name='Predicted GDP'
))
fig3.update_layout(
    title="Figure 3: GDP Forecast (Actual vs Predicted)",
    xaxis_title="Quarter",
    yaxis_title="GDP (INR)",
    template="plotly_white"
)
fig3.show()

# ------------------------
# FIGURE 4: SEASONAL PATTERNS
# ------------------------
fig4 = go.Figure()
fig4.add_trace(go.Bar(
    x=df_seasonal['name'], y=df_seasonal['avgGDP'], name='Average GDP'
))
fig4.update_layout(
    title="Figure 4: Seasonal GDP Patterns",
    xaxis_title="Quarter",
    yaxis_title="Average GDP",
    template="plotly_white"
)
fig4.show()

# ------------------------
# FIGURE 5: MODEL TRANSFERABILITY
# ------------------------
fig5 = go.Figure()
fig5.add_trace(go.Bar(
    x=df_transfer['model'], y=df_transfer['gdp'], name='GDP MAPE'
))
fig5.add_trace(go.Bar(
    x=df_transfer['model'], y=df_transfer['gva'], name='GVA MAPE'
))
fig5.update_layout(
    title="Figure 5: Model Transferability (GDP vs GVA MAPE)",
    xaxis_title="Model",
    yaxis_title="MAPE (%)",
    barmode="group",
    template="plotly_white"
)
fig5.show()
```
